If / For / While

```python
x = int(input('Primer valor: '))
y = int(input('Segundo valor: '))

print(f'¿Que operacion desea: ')
print(f' 1) Suma')
print(f' 2) Resta')
print(f' 3) Multiplicacion')
print(f' 4) Division')

operacion = int(input())

if operacion == 1:
  print(f'La respuesta es: {x+y}')
elif operacion == 2:
  print(f'La respuesta es: {x-y}')
elif operacion == 3:
  print(f'La respuesta es: {x*y}')
else:
  print(f'La respuesta es: {x/y}')
```

**Listas**

```python
calif = [4, 6, 9, 7, 5]

suma = 0
for i in range(len(calif)):
  #print(f'Posicion {i} Calificacion: {calif[i]}')
  suma = suma + calif[i]

prom = suma / len(calif)
print(prom)

op = 0
for e in calif:
  op += e

print(f'Promedio: {op/len(calif)}')
```

```python
cantidad = int(input('Cantidad de datos: '))

lista = []

for i in range(cantidad):
  lista.append(int(input('Value: ')))

print(lista)
#for e in lista:
#  print(e)
```
